# Zadanie 3
Na wykresie USD/VES w czasie zidentyfikuj i zaznacz kluczowe momenty za pomocą adnotacji (annotate).

Znajdź:
- Najwyższy kurs
- Najniższy kurs
- Największy wzrost (dzień do dnia)
- Największy spadek (dzień do dnia)

**Wskazówka:**

- użyj:  `.max()`, `.min()`, `.idxmax()`, `.idxmin()`
- użyj:  `pct_change()`


Dodaj do wykresu adnotacje:
- „Max rate”
- „Min rate”
- „Biggest increase”
- „Biggest drop”

Każda adnotacja powinna:
- wskazywać konkretny punkt (xy)
- mieć przesunięty opis (xytext) - trochę wyżej niż punkt
- zawierać strzałkę (arrowprops)

```python
# obszar wykresu
# Znajdź kluczowe momenty
# max
max_rate_value = ves_usd['Rate'].max()
date_max_rate = ves_usd['Rate'].idxmax()

#min
min_rate_value = ves_usd['Rate'].min()
date_min_rate = ves_usd['Rate'].idxmin()

# największy wzrost
biggest_increase_value = ves_usd['pct_change'].max()
biggest_increase_date = ves_usd['pct_change'].idxmax()

# największy spadek
biggest_drop_value = ves_usd['pct_change'].min()
biggest_drop_date = ves_usd['pct_change'].idxmin()

# wykres
fig, ax = plt.subplots(figsize=(16, 6))
ax.plot(ves_usd.index, ves_usd['Rate'], color='purple', label='VES to USD Rate')

# etykieta osi X
ax.set_xlabel('Data')
# etykieta osi y
ax.set_ylabel('Kurs (VES/USD)')

# tytuł wykresu
ax.set_title('Zależność Boliwara od Dolara Amerykańskiego z kluczowymi momentami')
ax.xaxis.set_tick_params(rotation=45)

# ustawienie zakresu dat
ax.set_xlim([ves_usd.index.min(), ves_usd.index.max()])

ax.grid(True)
ax.legend()

rate_range = ves_usd['Rate'].max() - ves_usd['Rate'].min()
offset_y = rate_range * 0.05 

# aby uniknac nachodzenia adnotacji
shift_y = offset_y * 2  

max_text_y = max_rate_value + offset_y
min_text_y = min_rate_value - offset_y
inc_text_y = ves_usd.loc[biggest_increase_date, 'Rate'] + offset_y
drop_text_y = ves_usd.loc[biggest_drop_date, 'Rate'] - offset_y

if date_max_rate == biggest_increase_date:
    inc_text_y += shift_y
if date_max_rate == biggest_drop_date:
    drop_text_y -= shift_y
if biggest_increase_date == biggest_drop_date:
    drop_text_y -= shift_y

# Adnotacja dla najwyzszego kursu
ax.annotate(f'Max rate: {max_rate_value:.2f}',
            xy=(date_max_rate, max_rate_value),
            xytext=(date_max_rate, max_text_y),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6),
            bbox=dict(boxstyle='round,pad=0.3', fc='yellow', ec='b', lw=1, alpha=0.6),
            ha='center')

# Adnotacja dla najnizszego kursu
ax.annotate(f'Min rate: {min_rate_value:.2f}',
            xy=(date_min_rate, min_rate_value),
            xytext=(date_min_rate, min_text_y),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6),
            bbox=dict(boxstyle='round,pad=0.3', fc='lightblue', ec='b', lw=1, alpha=0.6),
            ha='center')

# Adnotacja dla najwiekszego wzrostu
rate_at_biggest_increase = ves_usd.loc[biggest_increase_date, 'Rate']
ax.annotate(f'Biggest increase: {biggest_increase_value:.2%}',
            xy=(biggest_increase_date, rate_at_biggest_increase),
            xytext=(biggest_increase_date, inc_text_y),
            arrowprops=dict(facecolor='green', shrink=0.05, width=1, headwidth=6),
            bbox=dict(boxstyle='round,pad=0.3', fc='lightgreen', ec='g', lw=1, alpha=0.6),
            ha='center')

# Adnotacja dla najwiekszego spadku 
rate_at_biggest_drop = ves_usd.loc[biggest_drop_date, 'Rate']
ax.annotate(f'Biggest drop: {biggest_drop_value:.2%}',
            xy=(biggest_drop_date, rate_at_biggest_drop),
            xytext=(biggest_drop_date, drop_text_y),
            arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
            bbox=dict(boxstyle='round,pad=0.3', fc='salmon', ec='r', lw=1, alpha=0.6),
            ha='center')

plt.show()
```
