### Getting data type
We can use type() function.

```python
print(type(3.14)) # float
print(type({"apple", "banana", "cherry"})) # set
print(type([1, 2, 3, 4, 5])) # list
print(type((1, 2, 3, 4, 5))) # tuple
print(type({"name": "John", "age": 30, "city": "New York"})) # dict
```

## Casting
Casting in python is therefore done using constructor functions:

- int() - constructs an integer number from an integer literal, a float literal (by removing all decimals), or a string literal (providing the string represents a whole number)
- float() - constructs a float number from an integer literal, a float literal or a string literal (providing the string represents a float or an integer)
- str() - constructs a string from a wide variety of data types, including strings, integer literals and float literals

```python
x = int(1)   # x will be 1
y = int(2.8) # y will be 2

x = float(1)     # x will be 1.0
y = float(2.8)   # y will be 2.8
z = float("3")   # z will be 3.0
w = float("4.2") # w will be 4.2
```

### Lists
List are used to store multiple items in a single variable. List uses square brackets. List items are ordered (the items have defined order), changeable (we can change, add, remove values) and allow duplicates.

```python
list1 = ["Andrew", 3, True, 3.14]
print(list1)
print(type(list1)) # list
```

Access to list is simple. Lists are indexed, just need to referr to index number. Indexs start from 0 to n-1.

```python
print(list1[1:3]) # see 1 and 2 index
print(list1[0:4:2]) # see 0 and 2 index 
print(list1[:2]) # see first two elements
print(list1[2:]) # see from index 2 to the end
```

We can check if something is on list.

```python
if 3 in list1:
    print("3 is in the list")

if "Mark" not in list1:
    print("Mark is not in the list")
```

Other methods:
- sort() - list1.sort() - sort values in list
- append() - list1.append("orange) - add element on the end
- insert() - list1.insert(2, "blue") - insert blue on 2 index and moves values in list
- remove() - list1.remove(3) - remove item from list
- pop() - list1.pop(1) - remove value at index 1
- del:
  - del list 1 - delete list
  - del list1[0] - delete value on index 0

```python
list1.insert(2, "Hello") # insert "Hello" at index 2
print(list1)
```

### Tuples
Tuples are used to store collections of data. A tuple is collection which is ordered and unchangeable, allow duplacates. Tuples are also indexed.

```python
tuple1 = ("apple", "banana", "cherry", "apple", "cherry", 3, True, 3.14)
print(tuple1)
print(type(tuple1)) # tuple
```

Access to tuple is the same as list. Use indexes.

Add items is not that easy. Tuples are immutable. But we can change tuple to list or add two tuples into one.

```python
# convert list to tuple
list2 = list(tuple1)
print(list2)
print(type(list2)) # list

list2.append("orange") # add "orange" to the end of the list
print(list2)
tuple2 = tuple(list2) # convert list to tuple again
print(tuple2)
print(type(tuple2)) # tuple
```

```python
# two tuples into one
tuple2 = ("apple", "banana", "cherry")
y = ("orange", "melon",)
tuple2 += y

print(tuple2)
```

### Set
Set are used to store multiple items in a single variable. Set is unordered, but you can remove items and add new (no indexes). What is the most important **set don't allow duplicates**. Sets are written with curly brackets.

```python
set1 = {"Mark", 3, True, 3.14, "music"}
print(set1)
print(type(set1)) # set
```

There is no indexes so we need to use print(list_name) or simple loop.

```python
for element in set1:
    print(element)
```

Other methods:
- add() - set1.add("orange") - add element to the end
- update() - to add items from another set into current
- remove() - if value don't exsist, it will be error
- discard() - without error if  variable doesn't exists

```python
thisset = {"apple", "banana", "cherry"}
mylist = ["kiwi", "orange"]

thisset.update(mylist)

print(thisset)

thisset.discard("cucumber")
```

### Dictionary
Dictionaries are used to store data value in key:value pairs. Dictionaries are ordered (python 3.7, earlier not ordered), changeable and not allow duplcates. Dictionaries are written with curly brackets and have keys and values. 

```python
dict1 = {"name": "Jackob", "age": 30, "city": "New York", "year": 2000, "is_student": False}
print(dict1)
print(type(dict1)) # dict
```

Access items - we use keys or get method.

```python
print(dict1["name"]) # access value by key
print(dict1.get("age")) # access value by key using get() method
```

```python
# see dictionary keys
for x in dict1:
    print(x) # print keys
```

```python
# method 1 to see values
for x in dict1:
    print(dict1[x]) # print values

# method 2 to see values
for x in dict1.values():
    print(x) # print values
```

```python
"Jacob" in dict1.values() # check if value exists in dictionary
```

Dictionary methods:
- clear()	Removes all the elements from the dictionary
- copy()	Returns a copy of the dictionary
- fromkeys()	Returns a dictionary with the specified keys and value
- get()	Returns the value of the specified key
- items()	Returns a list containing a tuple for each key value pair
- keys()	Returns a list containing the dictionary's keys
- pop()	Removes the element with the specified key
- popitem()	Removes the last inserted key-value pair
- setdefault()	Returns the value of the specified key. If the key does not exist: insert the key, with the specified value
- update()	Updates the dictionary with the specified key-value pairs
- values()	Returns a list of all the values in the dictionary

```python
dict1.update({"country": "USA"}) # add new key-value pair to dictionary
print(dict1)

dict1.pop("year") # remove key "year" from dictionary
print(dict1)

dict1.update({"year": 2000}) # add key "year" back to dictionary
dict1
```

## Functions
Function is a block of code which only runs when it is called. Can return any data type as a result. Function avoids code repetition.
We use def functiona_name() to define a function.

```python
def my_function():
    print("Hello from a function")

my_function() # call the function
```

```python
def my_function2(nums):
    for num in nums:
        print(num)

my_function2([1, 2, 3, 4, 5]) # call the function with a list of numbers
```

## Function arguments
- information can be passed as arguments
- parameters are specified in brackets (), you can add as many params as you want
- a function must be called with the correct number of arguments, if we don't know how many it will be we can use *args and **kwargs

```python
def my_function3(name):
    print(f"Hello, {name} from a function")

my_function3("Alice") # call the function with a name
```

### *args, **kwargs
*** args** - function will receive tuple of arguments, it allows a function to accept any number of positional arguments

****kwargs** - function will receive a dictionary of arguments, it allows a function to accept any number of keyword arguments

We can mix natural arguments, *args and **kwargs.

```python
def casual_function(number1, number2):
    added = number1 + number2
    return added

result = casual_function(5, 10)
print("function result:", result)
print(type(result)) # int
```

```python
def arg_function(*args):
    print("arguments:", args) # print the arguments as a tuple
    total = 0
    for arg in args:
        total += arg
    return total

result = arg_function(1, 2, 3, 4, 5)
print("function result:", result)
print(type(result)) # int
```

```python
def arg_function(**kwargs):
    print("arguments:", kwargs) # print the arguments as a dictionary
    total = 0
    for key, value in kwargs.items():
        total += value
    return total

result = arg_function(a=1, b=2, c=3, d=4, e=5)
print("function result:", result)
print(type(result)) # int
```

```python
def mixed_function(title, *arg, **kwargs):
    print("arguments:", title, arg, kwargs)

result = mixed_function("User info", 1, "users", age=30, city="Manchester")
# title, (id, table)-tuple, {details}-dictionary
```

## JSON parsing - conver from JSON to Python
- json.loads() - parsing JSON for python dictionary
- json.dumps() - parsing Python object into JSON string

```python
import json

json_form = '{"name": "Mark", "age": 30, "city": "New York"}'
dict_form = json.loads(json_form)
print("dict form:", dict_form)
print("json form:", json_form)
```

```python
dict_form2 = {"name": "Monika", "age": 30}
json_form2 = json.dumps(dict_form2)

print("dict form:", dict_form2)
print("json form:", json_form2)
```
